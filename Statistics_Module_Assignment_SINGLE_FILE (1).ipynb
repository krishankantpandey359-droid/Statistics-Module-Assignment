**Dataset Columns Information**

PID = a unique identifier for the phone model

Blue = whether the phone has bluetooth support or not

Scr_w = width of the screen of the phone in centimetres (cm)

Int_Mem = internal memory of the phone measured in megabytes (MB)

Bty_Pwr = maximum energy stored by the phone's battery measured in
milli-Ampere-hours (mAh)

PC = resolution of the primary camera measued in megapixels (MP)

FC = resolution of the front camera measued in megapixels (MP)

RAM = random access memory available in the phone measured in gigabytes (GB)

Depth = depth of the mobile phone measured in centimetres (cm)

Weight = weight of the mobile phone measured in grams (g)

Price = selling price of the mobile phone in rupees

## Task 3 - Obtain the logical conditions for the features "Px_h" and "Px_w"

```python
# Get the feature names of the dataframe
print(df.columns.tolist())
```

```python
# Let's tackle these features: Px_h, Px_w
display(df[["Px_h", "Px_w"]].head())
```

```python
# Create Px as the total screen resolution
df["Px"] = df["Px_h"] * df["Px_w"]
print("Px feature created successfully.")
display(df[["Px_h", "Px_w", "Px"]].head())
```

```python
# Histogram of Px with mean and median
px_mean = df["Px"].mean()
px_median = df["Px"].median()

plt.figure(figsize=(9, 5))
plt.hist(df["Px"], bins=10, edgecolor="black")
plt.axvline(px_mean, linestyle="--", linewidth=2, label=f"Mean = {px_mean:,.0f}")
plt.axvline(px_median, linestyle=":", linewidth=2, label=f"Median = {px_median:,.0f}")
plt.title("Distribution of Total Screen Resolution (Px)")
plt.xlabel("Px")
plt.ylabel("Number of Phones")
plt.legend()
plt.show()

print(f"Mean Px = {px_mean:,.2f}")
print(f"Median Px = {px_median:,.2f}")
```

```python
# Good screen resolution: Px greater than or equal to the median
px_threshold = df["Px"].median()
con2 = df["Px"] >= px_threshold

print(f"Px threshold (median) = {px_threshold:,.2f}")
print("Phones satisfying con2:", int(con2.sum()))
```

Observations:

The features "Px_h" and "Px_w" are respectively the number of pixels in the phone screen in the vertical and horizontal axes.

We created a new feature called "Px" which is the product of the features "Px_h" and "Px_w".

The median has been selected as a threshold in this case.

In case it is too strict, we can choose the mean as a threshold.

# Task 9 - Study the variability of the features in the original data set

```python
# Calculate standard deviation / mean for all numerical features
# Exclude PID if it is non-numeric. Also avoid division by zero.
numeric_features = df.select_dtypes(include=np.number).columns

deviation_values = {}
for feature in numeric_features:
    mean_value = df[feature].mean()
    std_value = df[feature].std()
    deviation_values[feature] = np.nan if mean_value == 0 else std_value / mean_value

deviations = pd.DataFrame.from_dict(
    deviation_values, orient="index", columns=["Std_Dev_to_Mean"]
)
deviations.index.name = "Feature"

print("Standard deviation to mean ratio:")
display(deviations)
```

```python
# View the deviations series/table after sorting in descending order
deviations_sorted = deviations.sort_values(by="Std_Dev_to_Mean", ascending=False)
display(deviations_sorted)

print("Most variable numerical feature:",
      deviations_sorted.index[0])

print("Least variable numerical feature:",
      deviations_sorted.dropna().index[-1])
```
